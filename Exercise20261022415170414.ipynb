### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### Part 3：思考與分析（Questions）

1. 在函數內部，`scores` 的資料型態是什麼？其值為何？

`scores` 的資料型態是 tuple（元組），值為： (60, 70, 80)

因為函數定義中的 `*scores` 會將不定數量的位置參數收集成一個 tuple。

2. 在函數內部，`options` 的資料型態是什麼？其值為何？

`options` 的資料型態是 dict（字典），值為：

{
    "multiplier": 1.1,
    "bonus": 5
}

因為 `**options` 會將不定數量的關鍵字參數收集成字典。

3. `score * multiplier` 與 `score ** 2` 有什麼不同？請說明兩個運算子的意義。

- `*`：乘法運算，例如 `60 * 1.1` 的結果為 `66.0`。
- `**`：次方運算，例如 `71 ** 2` 的結果為 `5041`。

兩者用途不同，不能混為一談。

4. 為什麼函數使用 `return`，而不是只使用 `print()`？

`return` 可以將計算結果傳回呼叫函數的地方，讓其他程式繼續使用結果；`print()` 只是將內容顯示在畫面上。

例如：

result = analyze_scores(60, 70, 80)

print(result)

這樣才能把函數回傳的字典存進 `result` 變數。

5. 如果呼叫 `analyze_scores(60, 70, 80)`，但沒有提供 `settings`，程式會如何執行？請說明原因。

程式會使用預設值：

multiplier = 1

bonus = 0

因此調整後的分數仍為 `[60, 70, 80]`，總分為 `210`，平均為 `70.0`。

原因是：

options.get("multiplier", 1)

options.get("bonus", 0)

當字典中沒有指定的鍵時，就會使用第二個參數提供的預設值。

6. 請修改函數，讓回傳的字典額外包含調整後的最高分數，鍵值名稱為 `"max_score"`。

在原本的 return 字典中增加一個鍵值：

"max_score": max(adjusted) if adjusted else 0

這裡使用 max() 找出最高分。如果沒有分數，就回傳 0，避免對空串列使用 max() 而產生錯誤。

7. 如果沒有傳入任何分數，程式會回傳什麼結果？請解釋程式如何處理這種情況。

{
    'adjusted_scores': [],
    'squared_scores': [],
    'total': 0,
    'average': 0
}

原因如下：
- `adjusted` 是空串列 `[]`。
- `squared` 也是空串列 `[]`。
- `sum([])` 的結果為 `0`。
- `if adjusted else 0` 會在沒有分數時讓平均值回傳 `0。

1. 原本的 `*scores` 與 `scores_list` 有什麼不同？


會將串列傳給名為 `scores_list` 的關鍵字參數。

2. 為什麼原本的函數無法直接使用上述呼叫方式處理 `scores_list`？

原本的函數只有 `*scores` 和 `**options`，因此 `scores_list` 會被收集到 `options` 字典中，而不會自動成為 `scores` 裡的分數。

## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |

Python 中的 *、** 和 return 有不同用途。
1. `*` 可以表示乘法，也可以在函數定義中收集不定數量的位置參數，或在函數呼叫時解包串列等可迭代物件。
2. `**` 可以表示次方，也可以在函數定義中收集關鍵字參數，或在函數呼叫時解包字典。
3. `return` 可以將函數計算的結果傳回呼叫端，方便後續儲存和使用。

````python
def analyze_scores(*scores, **options):
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    if "scores_list" in options:
        scores = tuple(options["scores_list"])

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average,
        "max_score": max(adjusted) if adjusted else 0
    }
````
