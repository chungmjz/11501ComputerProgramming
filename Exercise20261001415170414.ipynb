1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。

十進位小數轉成二進位，可以一直乘以 2，並記錄每次的整數部分。

以 0.1 為例：
| 計算            | 整數部分 |
| ------------- | ---: |
| 0.1 × 2 = 0.2 |    0 |
| 0.2 × 2 = 0.4 |    0 |
| 0.4 × 2 = 0.8 |    0 |
| 0.8 × 2 = 1.6 |    1 |
| 0.6 × 2 = 1.2 |    1 |
| 0.2 × 2 = 0.4 |    0 |
| 0.4 × 2 = 0.8 |    0 |
| ……            |   …… |

所以得到：

0.1₁₀ = 0.0001100110011001100110011...₂

可以看到 `1100` 會不斷重複：

0.000110011001100110011001100...

          ↑──↑──↑──↑──↑──↑

因此，0.1 在二進位中是一個無限循環小數，無法用有限位數精確表示。

這也是 Python 浮點數計算會產生誤差的其中一個原因。


2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙

Python 一般使用 **IEEE 754 雙精度(double precision)**來儲存 `float`。

IEEE 754 雙精度浮點數總共使用 64 bits：

| 部分       |      位數 | 功能   |
| -------- | ------: | ---- |
| Sign     |   1 bit | 正負號  |
| Exponent | 11 bits | 指數   |
| Fraction | 52 bits | 小數部分 |
| 總計       | 64 bits |      |

因為電腦只能使用有限的位元，所以像 `0.1` 這種無限循環的二進位小數，必須截取有限位數來儲存。

也就是：

0.1

↓

二進位無限循環

↓

只能取有限位數

↓

取最接近的 IEEE 754 雙精度值

IEEE 754 使用的是 round to nearest, ties to even（最近值、平手取偶），也就是題目所說的「四捨六入五成雙」。

| 項目         | 意義                                          |
| ---------- | ------------------------------------------- |
| `max`      | Python `float` 能表示的最大值，約 `1.797693 × 10³⁰⁸` |

其中最重要的是：

mant_dig = 53

代表 IEEE 754 雙精度浮點數具有約 53 bits 的有效二進位精度。

4. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```

方法一：使用 `round()`   (`round(數值, 1)` 表示取到小數點後 1 位。)

方法二：使用 `math.isclose()` (如果目的是比較兩個浮點數是否足夠接近，比較推薦)

因為 `math.isclose()` 不要求兩個浮點數必須完全相等，而是判斷它們是否在合理的誤差範圍內。

5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。

不能一概而論。

如果是一般科學計算，IEEE 754 浮點數通常非常實用，因為它可以有效處理非常大或非常小的數值，而且計算速度快。

但是在金融計算中，如果要求金額必須精確，例如：

0.1 元 + 0.2 元 = 0.3 元

直接使用 `float` 可能產生誤差，因此不適合直接拿來處理需要精確十進位結果的金額。

Python 可以使用 `Decimal`：

因為 `Decimal` 是以十進位數字進行精確的十進位運算，更適合金融、會計等需要精確小數的情況。

````python
import sys
print(sys.float_info)
````

````python
print(0.1 + 0.2)
````

````python
print(round(0.1 + 0.2, 1))
````

````python
import math

print(math.isclose(0.1 + 0.2, 0.3))
````

````python
from decimal import Decimal

print(Decimal("0.1") + Decimal("0.2"))
````
