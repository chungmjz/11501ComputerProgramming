## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？

因為 Python 的 `int` 採用 Arbitrary Precision（任意精度）。

也就是說，不會像一般固定 32-bit 或 64-bit 整數那樣有固定的最大值。

2. Python 的 `int` 是否具有固定的位元數限制？

沒有固定的位元數上限。

但並不代表真的無限大。

3. 什麼是 Arbitrary Precision（任意精度）？

意思是：

整數所使用的儲存空間可以隨著數字變大而增加。

4. CPython 使用什麼物件來實作 `int`？

`PyLongObject`

5. Python 整數大小真正的限制是什麼？

主要受到：

可用記憶體、電腦資源和運算時間 限制。
 
所以「Unlimited Precision」比較準確的理解是：

沒有固定的精度上限，而不是實際上可以無限儲存。

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？

因為電腦使用二進位表示浮點數，而 `0.1`、`0.2` 這類十進位小數通常無法用有限的二進位小數精確表示。

2. 浮點數是否能精確表示所有小數？

不能。

3. 什麼是浮點數的精度（precision）？

精度（precision）是指浮點數能夠表示數值的細緻程度，也就是有效位數的能力。

4. 什麼是浮點數的範圍（range）？

範圍（range）是指可以表示的最小與最大數值範圍。

5. 精度與範圍有何不同？

Precision = 有多精準

Range     = 能表示多大／多小

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？

常見的是：

FP32、FP16、BF16

有些模型或硬體也會使用更低位元的格式。

2. 什麼是 FP64？

64-bit Floating Point

也就是：

64 位元浮點數 通常具有較高精度。

3. 什麼是 FP32？

32-bit Floating Point

常見於 AI 計算。

4. 什麼是 FP16？

16-bit Floating Point

比 FP32 使用更少記憶體，計算通常也更有效率。

5. 什麼是 BF16？

Brain Floating Point 16

也是 16-bit 浮點格式。

它保留與 FP32 相同的 exponent 位元數，因此具有較大的數值範圍，但有效精度低於 FP32。

6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？

因為 AI 模型需要處理大量的連續數值與矩陣運算。

浮點格式可以在：

記憶體使用量、計算速度、數值範圍、GPU 硬體效率

之間取得適合 AI 的平衡。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？

代表 Python `int 沒有固定的位元數上限，可以隨數字變大而使用更多儲存空間。

2. 它是否代表可以儲存無限大的整數？

不是。

實際上仍然受到：

記憶體、運算時間、電腦硬體

限制。

使用PyLongObject並以多個內部數值單位儲存大型整數，數字越大，可以配置更多空間。

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？

因為 Python `int 可以使用動態記憶體儲存更多位數。

2. 為什麼 `float` 卻存在範圍限制？

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。

Python int

→ 正確表示整數

→ 儲存空間可以增加

→ 正確性優先

FP16 / BF16

→ 固定較低位元數

→ 犧牲部分精度

→ 換取速度與記憶體效率

2. 各自的優點與缺點。

Python `int`

優點：

- 整數可以精確表示
- 可以表示非常大的整數

缺點：

- 大數字需要更多記憶體
- 數字越大，運算可能越慢

FP16 / BF16

優點：

- 使用較少記憶體
- GPU 運算效率高
- 適合大量矩陣運算


缺點：

- 精度有限
- 某些數值可能產生誤差

3. 為什麼大型語言模型通常選擇較低精度的表示方式？

因為大型模型有大量參數。

如果每個參數都使用高精度格式：

需要更多記憶體

↓

需要更大的 GPU 記憶體

↓

運算成本增加

↓

速度降低

使用 FP16、BF16 等格式，可以降低記憶體需求並提高硬體運算效率。

4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？

可能造成：

- 記憶體需求大幅增加
- GPU 記憶體壓力增加
- 運算成本提高
- 訓練／推論效率下降
- 能耗增加

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。

`x = 10**1000` 可以正常運作，是因為 Python 的 `int` 使用 Arbitrary Precision，可以隨著整數變大而配置更多記憶體，因此不受固定 32-bit 或 64-bit 整數大小的限制。不過它仍然受到電腦記憶體與運算時間的限制。

大型 AI 模型則通常使用 FP32、FP16 或 BF16，而不是任意精度整數。原因是 AI 模型具有非常大量的參數，如果每個參數都使用高精度或任意精度表示，會需要大量記憶體，而且運算速度較慢。FP16 和 BF16 只使用 16 bit，可以大幅降低記憶體使用量，並且能利用 GPU 的低精度運算能力提高效率。雖然低精度會犧牲部分數值精度，但對許多 AI 運算而言，這種誤差可以接受。因此 Python `int` 比較重視整數的正確性，而 AI 數值表示則在精度、記憶體與計算速度之間取得平衡。
